```python
 # Looping  Iteration Program
 #for Loop
 #Using Ranges
 for i in range(5):
     print(i)
 #Usins Data Types
 #Single Variable
 Fruits = ["Apple","Cherry","lichee","Pineapple","Watermelon"]
 for Fruit in Fruits:
     print(Fruit)
 #Double Variable
 Fruits = ["Apple","Cherry","lichee","Pineapple","Watermelon"]
 Price = [200,350,400,80,60]
 for Fruit,Amount in zip(Fruits,Price):
     print(f"{Fruit}:{Price}")
#while Loop
#Single While Loop
count = 0
while count < 5:
    print(count)
    count += 1
#Double While Loop
N = int(input("Enter a Number:"))
i = 1
while i <= N:
    print("*"*i)
    i += 1
j = 1
while j <= N:
    print("*"*(N-j))
    j += 1
#Loop Control Statement
#Break Statement
for i in range(1,11):
    if i == 5:
        break
    print(i)
#Continue Statement
Scores = [30,40,50,60,70]
Search = 50
for Score in Scores:
    if Score == 50:
        continue
    print(Score)
#Pass Statement
Employees = ["Sound","Bala","Chitra","Dinesh","Esha"]
for Employee in Employees:
    if Employee == "Chitra":
        pass
    else:
        print(Employee)
#Enumerate() with Index
Fruits = ["Apple","Cherry","lichee","Pineapple","Watermelon"]
for Index,Fruit in enumerate(Fruits):
    print(Index,Fruit)
#Zip() Parallel Iteration
Fruits = ["Apple","Cherry","lichee","Pineapple","Watermelon"]
Price = [200,350,400,80,60]
for Fruit,Amount in zip(Fruits,Price):
    print(Fruit,Amount)
```
