```python
#Python Operator
#Arithmetic operator
M = int(input("Enter M value"))
N = int(input("Enter N value"))
print("Addition:",M+N)
print("Subtraction:",M-N)
print("Multiplication:",M*N)
print("Division:",M/N)
print("Modulus:",M%N)
print("Exponent:",M**N)
print("Floor Division:",M//N)
#Assignment Operator
a = 10
print("Initial:",a)
a += 5
print("After Addition:",a)
a -= 5
print("After Subtraction:",a)
a *= 5
print("After Multiplication:",a)
a /= 5
print("After Division:",a)
#Comparision Operator
a = 10
b = 5
print("Equal:",a==b)
print("Not Equal:",a!=b)
print("Greater Than:",a>b)
print("Less Than:",a<b)
print("Greater Than Equal To:",a>=b)
print("Less Than Equal To:",a<=b)
#Logical Operator
a = 10
b = 5
print(a > b and b < a)
print(a > b or b < a)
print(not a > b)
# Idetity Operator
a = [1,2,3]
b = a
c = [1,2,3]
print(a is b)
print(a is c)
print(a is not b)
print(a is not c)
#Membership Operator
Name = ["Maha","Sandy","Hari","Jai"]
print("Maha" in Name)
print("Jai" not in Name)
print("Muthu" in Name)
print("Brevis" not in Name)
```
